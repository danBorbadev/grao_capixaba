Esse notebook vai servir para fazer a limpeza dos dados, observei que o csv de clientes veio formatado de uma forma não compátivel com UTF-8, por isso precisamos transformá-lo, antes de dar a carga no banco.

```python
import pandas as pd
```

```python
df = pd.read_csv('../dados_originais/clientes_crm.csv', encoding='latin1', sep=';')
```

```python
# Padronizar nomes e email da vase de clientes_crm
df ['Nome'] = df['Nome'].str.strip().str.title()
df ['Email'] = df['E-mail'].str.strip().str.title()
```

```python
df.to_csv(
    '../dados_limpos/clientes_crm.csv',
    encoding ='utf-8',
    sep=';',
    index = False
)

print('Arquivo tratado e salvo com sucesso')
```
